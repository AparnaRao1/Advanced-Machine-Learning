```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(1)

true_mu = 100
true_sigma = 5
n = 1000

data = np.random.normal(true_mu, true_sigma, n)

print("True mean:", true_mu)
print("True std:", true_sigma)
```

```python
mu_mle = np.mean(data)
sigma_mle = np.sqrt(np.mean((data - mu_mle) ** 2))

print("Estimated mean (MLE):", mu_mle)
print("Estimated std (MLE):", sigma_mle)
```

```python
def log_likelihood(mu, sigma, data):
    n = len(data)
    return -n*np.log(sigma*np.sqrt(2*np.pi)) - np.sum((data - mu)**2) / (2*sigma**2)
```

```python
mu_values = np.linspace(mu_mle - 3, mu_mle + 3, 100)
ll_values = [log_likelihood(mu, sigma_mle, data) for mu in mu_values]

plt.plot(mu_values, ll_values)
plt.xlabel("Mean (μ)")
plt.ylabel("Log-Likelihood")
plt.title("Log-Likelihood vs Mean")
plt.show()
```

```python
sigma_values = np.linspace(0.5, 5, 100)
ll_sigma = [log_likelihood(mu_mle, s, data) for s in sigma_values]

plt.plot(sigma_values, ll_sigma)
plt.xlabel("Standard Deviation (σ)")
plt.ylabel("Log-Likelihood")
plt.title("Log-Likelihood vs Std")
plt.show()
```

```python
np.random.seed(0)

sample_sizes = [10, 30, 50, 100, 500, 1000]
mu_mle_values = []
sigma_mle_values = []

for n in sample_sizes:
    data = np.random.normal(true_mu, true_sigma, n)
    mu_hat = np.mean(data)
    sigma_hat = np.sqrt(np.mean((data - mu_hat) ** 2))
    
    mu_mle_values.append(mu_hat)
    sigma_mle_values.append(sigma_hat)
```

```python
plt.plot(sample_sizes, mu_mle_values, marker='o')
plt.axhline(true_mu, linestyle='--')
plt.xlabel("Sample Size")
plt.ylabel("MLE of Mean")
plt.title("Mean MLE Convergence")
plt.show()
```

```python
plt.plot(sample_sizes, sigma_mle_values, marker='o')
plt.axhline(true_sigma, linestyle='--')
plt.xlabel("Sample Size")
plt.ylabel("MLE of Std")
plt.title("Std MLE Convergence")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# 1. Generate Data
np.random.seed(42)
true_mu = 5.0
true_std = 2.0
data = np.random.normal(true_mu, true_std, 1000)

# 2. Log-Likelihood Function for Normal Distribution
def normal_log_likelihood(mu, std, data):
    # We use scipy.stats.norm.logpdf for numerical stability
    return np.sum(norm.logpdf(data, loc=mu, scale=std))

# 3. Create a grid of mu and std values to evaluate
mu_vals = np.linspace(4.5, 5.5, 50)
std_vals = np.linspace(1.5, 2.5, 50)
mu_grid, std_grid = np.meshgrid(mu_vals, std_vals)

# Calculate LL for each combination
ll_surface = np.array([normal_log_likelihood(m, s, data) 
                       for m, s in zip(np.ravel(mu_grid), np.ravel(std_grid))])
ll_surface = ll_surface.reshape(mu_grid.shape)

# 4. Plotting
plt.figure(figsize=(8, 6))
cp = plt.contourf(mu_grid, std_grid, ll_surface, levels=20, cmap='viridis')
plt.colorbar(cp, label='Log-Likelihood')
plt.scatter(true_mu, true_std, color='red', marker='x', label='True Parameters')
plt.xlabel('$\mu$ (Mean)')
plt.ylabel('$\sigma$ (Std Dev)')
plt.title('Log-Likelihood Surface for Normal Distribution')
plt.legend()
plt.show()
```

```python
sample_sizes = [10, 50, 100, 500, 1000, 5000]
mu_estimates = []
std_estimates = []

for n in sample_sizes:
    sample = np.random.normal(true_mu, true_std, n)
    mu_estimates.append(np.mean(sample))
    std_estimates.append(np.std(sample)) # MLE of sigma

# Plotting Convergence
fig, ax1 = plt.subplots(figsize=(10, 5))

ax1.plot(sample_sizes, mu_estimates, label='Estimated $\mu$', marker='o', color='blue')
ax1.axhline(true_mu, color='blue', linestyle='--', alpha=0.5)
ax1.set_xlabel('Sample Size')
ax1.set_ylabel('Mean Estimate', color='blue')

ax2 = ax1.twinx()
ax2.plot(sample_sizes, std_estimates, label='Estimated $\sigma$', marker='s', color='green')
ax2.axhline(true_std, color='green', linestyle='--', alpha=0.5)
ax2.set_ylabel('Std Dev Estimate', color='green')

plt.title('Convergence of Normal MLE Parameters')
plt.show()
```
